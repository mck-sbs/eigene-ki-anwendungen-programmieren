Benötigte Module importieren und Datei laden. Die ersten Zeilen werden ausgegeben.

```python
import pandas as pd
import tensorflow as tf
from keras.models import Model
import matplotlib.pyplot as plt
import numpy as np
```

Modell laden

```python
model = tf.keras.models.load_model('transfer.keras', compile=False)
model.compile(loss='sparse_categorical_crossentropy', optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001), metrics=['accuracy']) 
```

```python
file = "../Data/geste.png"

img = tf.keras.utils.load_img(file,target_size=(300, 300,3))
img = np.array(img)

plt.figure()
plt.imshow(img)
plt.show()
```

```python
class_names = ['rock', 'paper', 'scissor']

img = tf.keras.applications.vgg19.preprocess_input(img)

pred = model.predict(img.reshape(1, 300, 300, 3))

print("prediction softmax:",pred)
print("prediction max:",pred.argmax())
print("prediction string:",class_names[pred.argmax()])
```
