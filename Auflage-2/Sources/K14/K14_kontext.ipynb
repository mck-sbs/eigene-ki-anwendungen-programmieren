```python
from openai import OpenAI

API_KEY = "..."
```

```python
instructions = "Zu jedem Sternzeichen, zu dem du gefragt wirst, gibst du kurze, aber lustige Antworten. Wenn du nach etwas anderem als zu Sternzeichen gefragt wirst, sage, dass dich das nicht interessiert."
msg = "Wie sind Fische so?"
model = "gpt-4.1"

client = OpenAI(api_key=API_KEY)

response = client.responses.create(model=model, instructions=instructions, input=msg)
print(response.output_text)
```

```python
from google import genai
from google.genai import types

API_KEY = "..."

instructions = "Zu jedem Sternzeichen, zu dem du gefragt wirst, gibst du kurze, aber lustige Antworten. Wenn du nach etwas anderem als zu Sternzeichen gefragt wirst, sage, dass dich das nicht interessiert."
msg = "Wie sind Fische so?"

model = "gemini-2.0-flash"

config=types.GenerateContentConfig(system_instruction=instructions) 
client = genai.Client(api_key=API_KEY)

response = client.models.generate_content(model=model, contents=msg, config=config)
print(response.text)
```
