```python
from openai import OpenAI

API_KEY = "..."

msg = "Lebst du?"
tmp = 1.2
model = "gpt-4.1"

client = OpenAI(api_key=API_KEY)

for i in range(5):
    response = client.responses.create(model=model, input=msg, temperature=tmp)
    print("*", response.output_text)
```

```python
from google import genai
from google.genai import types

API_KEY = "..."

msg = "Lebst du?"
tmp = 1.2
model = "gemini-2.0-flash"

config=types.GenerateContentConfig(temperature=tmp) 
client = genai.Client(api_key=API_KEY)

for i in range(5):
    response = client.models.generate_content(model=model, contents=msg, config=config)
    print(response.text)
```
