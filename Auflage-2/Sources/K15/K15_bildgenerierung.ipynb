```python
n = 1
size = "1024x1024"
prompt = "Dinosaurier im Weltall"
model = "dall-e-3"

client = OpenAI(api_key=API_KEY)
response = client.images.generate(model=model , prompt=prompt, size=size)

print(response)
```

```python
# Bitte in Colab ausführen
from google import genai
from google.genai import types
from PIL import Image
from io import BytesIO
from IPython.display import display 

API_KEY = "..."

numberOfImages = 1
aspectRatio = "1:1"
personGeneration = "DONT_ALLOW"
model = "imagen-3.0-generate-002"
prompt = "Dinosaurier im Weltall"

config=types.GenerateImagesConfig(number_of_images=numberOfImages, aspectRatio=aspectRatio,
                                  personGeneration=personGeneration)

client = genai.Client(api_key=API_KEY)

response = client.models.generate_images(model=model, prompt=prompt, config=config)

image = Image.open(BytesIO(response.generated_images[0].image.image_bytes))
display(image) 
```
