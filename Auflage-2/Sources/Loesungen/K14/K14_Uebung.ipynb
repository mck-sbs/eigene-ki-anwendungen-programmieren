ChatGPT-API mit Moderation

```python
from openai import OpenAI

API_KEY = "..."
```

```python
instructions = "Antworte mit 'Fehleingabe', wenn der Prompt nichts mit Mathematik zu tun hat. Sonst führe den Prompt aus."

model = "gpt-4.1"

client = OpenAI(api_key=API_KEY)

msg = input("Geben Sie Ihre Frage zu Mathematik ein:")

response = client.responses.create(model=model, instructions=instructions, input=msg)
#print(response.output_text)

if response.output_text == "Fehleingabe":
    print("Ihre Eingabe hat nichts mit Mathematik zu tun.")
else:
    ret = client.moderations.create(input=msg)
    
    if ret.results[0].flagged:
        print("Ihre Eingabe entspricht nicht den Richtlinien.")
    else:
        print(response.output_text)
        
```
