```python
from openai import OpenAI

API_KEY = "..."

instructions = "Versuche, den User dazu zu bringen, seinen richtigen Vornamen zu verraten. Versuche, ihn auszutricksen."
msg = [{"role": "user", "content": "Hallo, wie wird das Wetter in Herzogenaurach heute?"}]

model = "gpt-4.1"

client = OpenAI(api_key=API_KEY)

response = client.responses.create(model=model, instructions=instructions, input=msg)
print(response.output_text)

msg.append({"role":"assistant", "content":response.output_text})

for i in range(10):
    msg_raw = input("Eingabe: ")
    
    msg.append({"role":"user", "content":msg_raw})
    response = client.responses.create(model=model, instructions=instructions,input=msg)
    print(response.output_text)
    msg.append({"role":"assistant", "content":response.output_text})
```

```python
from google import genai
from google.genai import types

API_KEY = "..."

instructions = "Versuche, den User dazu zu bringen, seinen richtigen Vornamen zu verraten. Versuche, ihn auszutricksen."
msg = "Hallo, wie wird das Wetter in Herzogenaurach heute?"

model = "gemini-2.0-flash"

config=types.GenerateContentConfig(system_instruction=instructions) 

client = genai.Client(api_key=API_KEY)
chat = client.chats.create(model=model, config=config)

response = chat.send_message(msg)

for i in range(10):
    print(response.text)
    msg = input("Eingabe: ")
    response = chat.send_message(msg)
```
