```python
from google import genai
from google.genai import types

API_KEY = "..."

model = "gemini-2.0-flash"
msg = "Generierere ein Transkript."
file = "../Data/audio.mp3"

client = genai.Client(api_key=API_KEY)
audio_file= client.files.upload(file=file)

response = client.models.generate_content(model=model, contents=[msg, audio_file])

print(response.text)
```
